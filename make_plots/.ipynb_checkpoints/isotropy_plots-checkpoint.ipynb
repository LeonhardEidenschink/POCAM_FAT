```python
import itertools
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
import json

import plot_utils
```

```python
pocam = 3
hem = 11
emitter = 'LMG405'
```

```python
file_path = f'../output/batch1/pocam_{pocam:03d}/hem_{hem:.0f}/{emitter}/isotropy.json'
```

```python
with open(file_path, "r") as f:
    data = json.load(f)
```

```python
d = data["meas_data"][0]
Z = np.array(d["z_values"])

x = np.linspace(d["x_min"], d["x_max"], d["x_bins"])
y = np.linspace(d["y_min"], d["y_max"], d["y_bins"])
X, Y = np.meshgrid(x, y)

fig = plt.figure(figsize=plot_utils.set_size('wide', fraction=0.7, subplots=(1, 1)))
ax = fig.add_subplot(111, projection="mollweide")

# Mollweide expects radians → you're already in radians
c = ax.pcolormesh(X, Y, Z, cmap = cm.coolwarm)

ax.set_title('Angular Emission Profile', y = 1.3)
fig.colorbar(c, ax=ax, label=d["z_label"])
```
